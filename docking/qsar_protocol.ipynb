```python
from rdkit import Chem
from rdkit.Chem import Draw
from rdkit.Chem.Draw import rdMolDraw2D

# Create two example molecules
mol1 = Chem.MolFromSmiles('CCO')
mol2 = Chem.MolFromSmiles('CCOC(=O)CC')

# Define the substructure pattern
substructure = Chem.MolFromSmarts('O')

# Get the substructure matches
matches_mol1 = mol1.GetSubstructMatches(substructure)
matches_mol2 = mol2.GetSubstructMatches(substructure)

# Prepare molecules for drawing
mol1 = rdMolDraw2D.PrepareMolForDrawing(mol1)
mol2 = rdMolDraw2D.PrepareMolForDrawing(mol2)

# Draw the molecules with highlighted substructure
drawer = rdMolDraw2D.MolDraw2DCairo(400, 200)
drawer.DrawMolecule(mol1, highlightAtoms=matches_mol1[0])
drawer.FinishDrawing()
image1 = drawer.GetDrawingText()

drawer = rdMolDraw2D.MolDraw2DCairo(400, 200)
drawer.DrawMolecule(mol2, highlightAtoms=matches_mol2[0])
drawer.FinishDrawing()
image2 = drawer.GetDrawingText()

# Save the images as PNG files
with open('mol1.png', 'wb') as f1, open('mol2.png', 'wb') as f2:
    f1.write(image1)
    f2.write(image2)
```

```python
from rdkit import Chem
from rdkit import DataStructs

# Create two example fingerprints (Morgan fingerprints)
mol1 = Chem.MolFromSmiles('CC1C=CC=CCCC=CC=CC=CC=CC(CC2C(C(CC(O2)(CC(C(CCC(CC(CC(CC(=O)OC(C(C1O)C)C)O)O)O)O)O)O)O)C(=O)O)OC3C(C(C(C(O3)C)O)N)O') # heptenene
mol2 = Chem.MolFromSmiles('CC1=CCC(CC1)C(=C)C')

fp1 = Chem.RDKFingerprint(mol1)
fp2 = Chem.RDKFingerprint(mol2)

# Calculate Tanimoto similarity
similarity = DataStructs.CosineSimilarity(fp1, fp2)
print(f"Tanimoto similarity: {similarity}")
```

```python
from rdkit import Chem
from rdkit.Chem import rdFMCS

# Create two example molecules
mol1 = Chem.MolFromSmiles('CC=CCCC')
mol2 = Chem.MolFromSmiles('CC1=CCC(CC1)C(=C)C')

# Find the maximum common substructure
mcs = rdFMCS.FindMCS([mol1, mol2])

# Convert the MCS query to a molecule
mcs_mol = Chem.MolFromSmarts(mcs.smartsString)

# Get the substructure matches
matches_mol1 = mol1.GetSubstructMatches(mcs_mol)
matches_mol2 = mol2.GetSubstructMatches(mcs_mol)

# Print the matches
print(f"Substructure matches in mol1: {matches_mol1}")
print(f"Substructure matches in mol2: {matches_mol2}")
```

```python
from rdkit import Chem
from rdkit.Chem import Descriptors

# Create a molecule object
smiles = "CC1=CCC(CC1)C(=C)C"
mol = Chem.MolFromSmiles(smiles)

# Calculate logP
logP = Descriptors.MolLogP(mol)

print("LogP:", logP)
```
